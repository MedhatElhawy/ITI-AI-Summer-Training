## 1. Why Object-Oriented Programming?

As programs grow, managing data and the functions that operate on that data separately becomes hard to maintain. Object-Oriented Programming (OOP) solves this by bundling **data** (attributes) and **behavior** (methods) into a single unit called an **object**.

### Core OOP Principles
- **Encapsulation** — bundling data and methods together, and controlling access to them
- **Inheritance** — creating new classes based on existing ones
- **Polymorphism** — the same interface can be used with different underlying data types
- **Abstraction** — hiding complex implementation details behind a simple interface

Let's see the difference between a procedural and an object-oriented approach.


```python
# Procedural approach: data and logic are separate
student_name = "Ahmed"
student_grade = 85

def print_student_info(name, grade):
    print(f"Student: {name}, Grade: {grade}")

print_student_info(student_name, student_grade)
```

```python
# Object-oriented approach: data and logic live together
class Student:
    def __init__(self, name, grade):
        self.name = name
        self.grade = grade

    def print_info(self):
        print(f"Student: {self.name}, Grade: {self.grade}")

student1 = Student("Ahmed", 85)
student1.print_info()
```

## 2. Classes and Objects

A **class** is a blueprint for creating objects. An **object** (or instance) is a concrete realization of that blueprint, with its own data.

- `class` defines the blueprint.
- `self` refers to the specific instance calling the method.
- Each object has its own copy of instance attributes.


```python
class Car:
    def __init__(self, brand, model, year):
        self.brand = brand
        self.model = model
        self.year = year

    def display_info(self):
        print(f"{self.year} {self.brand} {self.model}")

car1 = Car("Toyota", "Corolla", 2022)
car2 = Car("Honda", "Civic", 2023)

car1.display_info()
car2.display_info()

print(car1 is car2)  # Different objects, different memory addresses
```

### Extra Example: Multiple Objects, Independent State

```python
class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages
        self.current_page = 0

    def read(self, pages_to_read):
        self.current_page = min(self.current_page + pages_to_read, self.pages)
        print(f"'{self.title}': now on page {self.current_page}/{self.pages}")

book1 = Book("Clean Code", "Robert Martin", 464)
book2 = Book("Fluent Python", "Luciano Ramalho", 792)

book1.read(50)
book1.read(80)
book2.read(120)  # book2's progress is completely independent of book1
```

## 3. The `__init__` Constructor

`__init__` is a special method automatically called when a new object is created. It's used to initialize the object's attributes.

### Instance Attributes vs Class Attributes
- **Instance attributes** belong to a specific object (defined with `self.attribute`).
- **Class attributes** are shared by all instances of the class.


```python
class Employee:
    company_name = "700 Apps"  # Class attribute — shared by all instances

    def __init__(self, name, salary):
        self.name = name       # Instance attribute
        self.salary = salary   # Instance attribute

emp1 = Employee("Sara", 8000)
emp2 = Employee("Omar", 9500)

print(emp1.company_name, emp2.company_name)  # Same value, shared
print(emp1.name, emp1.salary)
print(emp2.name, emp2.salary)

# Changing a class attribute via the class affects all instances
Employee.company_name = "700 Apps Egypt"
print(emp1.company_name, emp2.company_name)
```

### Extra Example: Class Attribute as a Shared Counter

```python
class Robot:
    population = 0  # class attribute tracking how many robots exist

    def __init__(self, name):
        self.name = name
        Robot.population += 1

    @classmethod
    def how_many(cls):
        return cls.population

r1 = Robot("R2D2")
r2 = Robot("C3PO")
r3 = Robot("BB8")

print(f"Robots created: {Robot.how_many()}")
print(r1.population, r2.population, r3.population)  # all see the same shared value
```

## 4. Instance Methods vs Class Attributes

Instance methods always take `self` as their first parameter, giving them access to the instance's attributes and other methods.


```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print(f"Deposited {amount}. New balance: {self.balance}")

    def withdraw(self, amount):
        if amount > self.balance:
            print("Insufficient funds.")
        else:
            self.balance -= amount
            print(f"Withdrew {amount}. New balance: {self.balance}")

account = BankAccount("Mona", 1000)
account.deposit(500)
account.withdraw(2000)
account.withdraw(300)
```

### Extra Example: A Method That Uses Another Method

```python
class ShoppingCart:
    def __init__(self):
        self.items = []

    def add_item(self, name, price):
        self.items.append({"name": name, "price": price})
        print(f"Added {name} (${price})")

    def total(self):
        return sum(item["price"] for item in self.items)

    def checkout(self):
        total = self.total()  # calling another instance method
        print(f"Checking out {len(self.items)} items. Total: ${total:.2f}")
        self.items = []  # clear cart after checkout

cart = ShoppingCart()
cart.add_item("Keyboard", 25.5)
cart.add_item("Mouse", 12.0)
cart.checkout()
```

## 5. Encapsulation

Encapsulation restricts direct access to some of an object's components, which is a way of preventing accidental modification of data.

Python uses naming conventions (it does not enforce true private access like Java or C++):

| Convention | Meaning | Example |
|---|---|---|
| `name` | Public — accessible anywhere | `self.name` |
| `_name` | Protected — internal use, convention only | `self._name` |
| `__name` | Private — name-mangled by Python | `self.__name` |


```python
class Account:
    def __init__(self, owner, balance):
        self.owner = owner
        self._pin = "1234"        # protected (convention)
        self.__balance = balance  # private (name-mangled)

    def get_balance(self):
        return self.__balance

    def set_balance(self, amount):
        if amount >= 0:
            self.__balance = amount
        else:
            print("Balance cannot be negative.")

acc = Account("Youssef", 5000)
print(acc.get_balance())

# Direct access to private attribute fails as expected
try:
    print(acc.__balance)
except AttributeError as e:
    print("Error:", e)

# Python actually renames it internally:
print(acc._Account__balance)  # name mangling demonstration
```

### Extra Example: Using `property` for Controlled Access

A more Pythonic way to enforce validation instead of plain getter/setter methods is the `@property` decorator.

```python
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius

    @property
    def celsius(self):
        return self._celsius

    @celsius.setter
    def celsius(self, value):
        if value < -273.15:
            raise ValueError("Temperature below absolute zero is not possible.")
        self._celsius = value

    @property
    def fahrenheit(self):
        return self._celsius * 9/5 + 32

temp = Temperature(25)
print(temp.celsius, temp.fahrenheit)

temp.celsius = 30       # looks like attribute access, but runs the setter
print(temp.fahrenheit)

try:
    temp.celsius = -500
except ValueError as e:
    print("Error:", e)
```

## 6. Inheritance

Inheritance allows a class (child/subclass) to acquire attributes and methods from another class (parent/superclass), promoting code reuse.

Use `super()` to call the parent class's constructor or methods.


```python
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        print(f"{self.name} makes a sound.")

class Dog(Animal):
    def __init__(self, name, breed):
        super().__init__(name)   # call parent constructor
        self.breed = breed

    def speak(self):             # method overriding
        print(f"{self.name} the {self.breed} barks.")

class Cat(Animal):
    def speak(self):
        print(f"{self.name} meows.")

animals = [Dog("Rex", "Labrador"), Cat("Whiskers"), Animal("Generic")]
for a in animals:
    a.speak()
```

### Extra Example: Multi-level Inheritance

```python
class Vehicle:
    def __init__(self, brand):
        self.brand = brand

    def info(self):
        print(f"Vehicle brand: {self.brand}")

class Car(Vehicle):
    def __init__(self, brand, doors):
        super().__init__(brand)
        self.doors = doors

    def info(self):
        super().info()  # reuse the parent method, then extend it
        print(f"Number of doors: {self.doors}")

class SportsCar(Car):
    def __init__(self, brand, doors, top_speed):
        super().__init__(brand, doors)
        self.top_speed = top_speed

    def info(self):
        super().info()
        print(f"Top speed: {self.top_speed} km/h")

my_car = SportsCar("Ferrari", 2, 340)
my_car.info()
```

## 7. Polymorphism

Polymorphism means "many forms." The same method name behaves differently depending on the object that calls it — as seen in the loop above, `speak()` produces different output for `Dog`, `Cat`, and `Animal`.

Python also supports polymorphism through **duck typing**: "If it walks like a duck and quacks like a duck, it's a duck." We don't need explicit interfaces — any object that implements the expected method can be used.


```python
class Circle:
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return 3.14159 * self.radius ** 2

class Square:
    def __init__(self, side):
        self.side = side

    def area(self):
        return self.side ** 2

shapes = [Circle(5), Square(4)]
for shape in shapes:
    print(f"{shape.__class__.__name__} area: {shape.area():.2f}")
```

### Extra Example: Polymorphism with a Common Function

```python
class Triangle:
    def __init__(self, base, height):
        self.base = base
        self.height = height

    def area(self):
        return 0.5 * self.base * self.height

def print_total_area(shapes):
    """Works with ANY object that has an area() method — true polymorphism."""
    total = sum(shape.area() for shape in shapes)
    print(f"Total area: {total:.2f}")

shapes = [Circle(3), Square(5), Triangle(6, 4)]
print_total_area(shapes)
```

## 8. Special (Dunder) Methods

"Dunder" (double underscore) methods let your objects integrate naturally with Python's built-in syntax and functions (`print()`, `+`, `len()`, comparisons, etc.).

Common dunder methods:
| Method | Purpose |
|---|---|
| `__init__` | Constructor |
| `__str__` | Human-readable string (used by `print()`) |
| `__repr__` | Developer-facing representation |
| `__len__` | Enables `len(obj)` |
| `__eq__` | Enables `obj1 == obj2` |
| `__add__` | Enables `obj1 + obj2` |


```python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __str__(self):
        return f"Vector({self.x}, {self.y})"

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __eq__(self, other):
        return self.x == other.x and self.y == other.y

    def __len__(self):
        return int((self.x ** 2 + self.y ** 2) ** 0.5)

v1 = Vector(2, 3)
v2 = Vector(4, 1)

print(v1)              # uses __str__
print(v1 + v2)          # uses __add__
print(v1 == Vector(2, 3))  # uses __eq__
print(len(v1))          # uses __len__
```

### Extra Example: More Dunder Methods (`__lt__`, `__repr__`, `__getitem__`)

```python
class Playlist:
    def __init__(self, songs):
        self.songs = songs

    def __repr__(self):
        return f"Playlist({self.songs!r})"

    def __len__(self):
        return len(self.songs)

    def __getitem__(self, index):
        return self.songs[index]

    def __lt__(self, other):
        return len(self) < len(other)

p1 = Playlist(["Song A", "Song B"])
p2 = Playlist(["Song X", "Song Y", "Song Z"])

print(p1)             # uses __repr__
print(len(p2))         # uses __len__
print(p1[0])           # uses __getitem__
print(p1 < p2)         # uses __lt__

for song in p2:        # __getitem__ also enables iteration
    print("-", song)
```

## 9. Class Methods and Static Methods

- **Instance methods** (`self`) operate on a specific object.
- **Class methods** (`@classmethod`, `cls`) operate on the class itself — often used as alternative constructors.
- **Static methods** (`@staticmethod`) don't need access to the instance or class — they're utility functions grouped inside the class for organization.


```python
class Pizza:
    def __init__(self, toppings):
        self.toppings = toppings

    @classmethod
    def margherita(cls):
        return cls(["mozzarella", "tomato"])  # alternative constructor

    @staticmethod
    def is_valid_topping(topping):
        allowed = ["mozzarella", "tomato", "pepperoni", "mushroom"]
        return topping in allowed

pizza1 = Pizza.margherita()
print(pizza1.toppings)

print(Pizza.is_valid_topping("pepperoni"))
print(Pizza.is_valid_topping("pineapple"))
```

### Extra Example: Class Method as a Factory From Different Data Sources

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    @classmethod
    def from_dict(cls, data):
        return cls(data["name"], data["salary"])

    @classmethod
    def from_csv_line(cls, line):
        name, salary = line.split(",")
        return cls(name.strip(), float(salary.strip()))

    def __repr__(self):
        return f"Employee({self.name}, {self.salary})"

emp1 = Employee.from_dict({"name": "Laila", "salary": 12000})
emp2 = Employee.from_csv_line("Hassan, 9800")

print(emp1)
print(emp2)
```

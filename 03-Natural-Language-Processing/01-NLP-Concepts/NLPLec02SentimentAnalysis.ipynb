# Sentiment Analysis

**Sentiment Analysis** is an NLP technique used to identify the **emotion or opinion** expressed in a text.

In simple words:

> It tells us whether a text is **Positive, Negative, or Neutral**.

## Example

```text
"I love this movie!"
```

→ **Positive** 😊

```text
"I hate this movie."
```

→ **Negative** 😡

```text
"The movie was released yesterday."
```

→ **Neutral** 😐

## Common Applications

* **Product Reviews** → Determine whether customers are happy or unhappy.
* **Social Media** → Analyze people's opinions about a product, service, or topic.
* **Customer Feedback** → Identify positive and negative feedback.
* **Movie Reviews** → Determine whether people liked or disliked a movie.

## Example Dataset

| Review                    | Sentiment |
| ------------------------- | --------- |
| I love this product       | Positive  |
| This product is terrible  | Negative  |
| The product arrived today | Neutral   |

The goal of **Sentiment Analysis** is to automatically predict the `Sentiment` based on the text in the `Review` column.

## Sentiment Analysis vs NER vs POS

| NLP Task               | What does it identify?                               |
| ---------------------- | ---------------------------------------------------- |
| **POS Tagging**        | The grammatical type of each word                    |
| **NER**                | Named entities such as people, companies, and places |
| **Sentiment Analysis** | The emotion or opinion expressed in the text         |

### Simple Difference

**POS → What type of word is it?**

**NER → What entity is it?**

**Sentiment Analysis → What is the opinion or emotion?**


# Polarity Score

**Polarity Score** is a numerical value used in **Sentiment Analysis** to measure whether a text expresses a **positive or negative sentiment**.

The score usually ranges from:

```text
-1  ← Negative     0     Positive →  +1
```

## Interpretation

| Polarity Score | Sentiment     |
| -------------: | ------------- |
|           `-1` | Very Negative |
|         `-0.5` | Negative      |
|            `0` | Neutral       |
|          `0.5` | Positive      |
|            `1` | Very Positive |

### Example

```text
"I love this product!"
```

→ Polarity ≈ **+0.8**

This indicates a **strong positive sentiment**.

```text
"I hate this product."
```

→ Polarity ≈ **-0.8**

This indicates a **strong negative sentiment**.

```text
"The product arrived today."
```

→ Polarity ≈ **0**

This indicates a **neutral sentiment**.

## Important

A polarity score measures the **direction and strength of sentiment**:

* **Negative value** → negative sentiment
* **Positive value** → positive sentiment
* **Close to zero** → neutral or weak sentiment


# Lexicon in NLP

A **Lexicon** is a collection of words with information about their meaning, sentiment, or other linguistic properties.

In **Sentiment Analysis**, a **sentiment lexicon** is a list of words associated with positive or negative sentiment.

## Example

| Word      | Sentiment Score |
| --------- | --------------: |
| excellent |            +1.0 |
| good      |            +0.7 |
| happy     |            +0.8 |
| okay      |               0 |
| bad       |            -0.7 |
| terrible  |            -1.0 |
| hate      |            -0.9 |

For example:

```text
"I love this movie"
```

The lexicon may contain:

```text
love → +0.8
```

So the word **"love"** contributes a positive sentiment.

## How Does It Work?

A lexicon-based sentiment analysis approach generally:

1. Splits the text into words/tokens.
2. Looks up each word in the sentiment lexicon.
3. Gets the sentiment score for each word.
4. Combines the scores to estimate the sentiment of the whole text.

### Example

```text
"I love this movie but the ending is terrible."
```

The lexicon might give:

```text
love     → +0.8
terrible → -1.0
```

The overall sentiment depends on how the method combines these scores.

## Lexicon vs Polarity Score

| Concept            | Meaning                                                      |
| ------------------ | ------------------------------------------------------------ |
| **Lexicon**        | A dictionary/list containing sentiment information for words |
| **Polarity Score** | A numerical score representing the sentiment of a text       |

### Simple Example

**Lexicon:**

```text
love → +0.8
hate → -0.8
```

**Text:**

```text
"I love this product."
```

**Polarity:**

```text
+0.8
```

> **Lexicon = the dictionary of sentiment words**
> **Polarity Score = the resulting sentiment value**


# TextBlob and VADER

**TextBlob** and **VADER** are two popular tools used for **Sentiment Analysis** in NLP.

Both can analyze text and determine whether the sentiment is **Positive, Negative, or Neutral**.

---

## 1. TextBlob

**TextBlob** is a Python library that provides simple NLP functionality, including sentiment analysis.

It uses a **lexicon-based approach** to calculate sentiment.

### Example

```python
from textblob import TextBlob

text = "I love this movie!"

blob = TextBlob(text)

print(blob.sentiment)
```

Output:

```text
Sentiment(polarity=0.625, subjectivity=0.6)
```

### Polarity

The **polarity** score ranges from:

```text
-1  ← Negative     0     Positive →  +1
```

* `-1` → Very Negative
* `0` → Neutral
* `+1` → Very Positive

### Subjectivity

TextBlob also provides a **subjectivity score**:

```text
0 → Objective
1 → Subjective
```

For example:

```text
"The Earth is round."
```

→ More **objective**

```text
"I think this movie is amazing."
```

→ More **subjective**

---

# 2. VADER

**VADER** stands for:

> **Valence Aware Dictionary and sEntiment Reasoner**

VADER is a lexicon and rule-based sentiment analysis tool, especially designed for **social media and short informal text**.

It can handle things such as:

* Capitalization
* Exclamation marks
* Emojis
* Punctuation
* Intensifiers
* Negation

### Example

```python
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

analyzer = SentimentIntensityAnalyzer()

text = "I LOVE this movie!!!"

scores = analyzer.polarity_scores(text)

print(scores)
```

Possible output:

```text
{
    'neg': 0.0,
    'neu': 0.25,
    'pos': 0.75,
    'compound': 0.6696
}
```

---

## VADER Scores

VADER returns four scores:

| Score      | Meaning                 |
| ---------- | ----------------------- |
| `neg`      | Negative sentiment      |
| `neu`      | Neutral sentiment       |
| `pos`      | Positive sentiment      |
| `compound` | Overall sentiment score |

The **compound score** ranges from:

```text
-1 ← Very Negative      0      Very Positive → +1
```

---

# TextBlob vs VADER

| Feature           | TextBlob      | VADER                |
| ----------------- | ------------- | -------------------- |
| Type              | Lexicon-based | Lexicon + Rule-based |
| Polarity          | ✅             | ✅                    |
| Subjectivity      | ✅             | ❌                    |
| `compound` score  | ❌             | ✅                    |
| Emojis            | Limited       | ✅                    |
| Capitalization    | Limited       | ✅                    |
| Exclamation marks | Limited       | ✅                    |
| Social media      | Good          | **Very Good**        |
| Easy to use       | **Very Easy** | Easy                 |

## When Should You Use Each?

### Use TextBlob when:

* You want a **simple introduction** to sentiment analysis.
* You need **polarity and subjectivity**.
* Your text is relatively clean.

### Use VADER when:

* You are analyzing **tweets, comments, or social media**.
* Your text contains **emojis, slang, capitalization, or !!!**.
* You want a more detailed sentiment score.

### Simple Summary

**TextBlob:**

> "How positive/negative and subjective is this text?"

**VADER:**

> "How positive/negative is this text, considering things like punctuation, capitalization, emojis, and emphasis?"

# Pre-trained Transformer Models

**Pre-trained Transformer Models** are NLP models that have already been trained on a **large amount of text data** before being used for a specific task.

Instead of training a model from scratch, we can use an existing **pre-trained model** and apply it to our NLP task.

---

## What Does "Pre-trained" Mean?

**Pre-trained** means the model has already learned patterns and relationships from a huge amount of text.

For example, a model may learn:

* Word meanings
* Grammar
* Context
* Relationships between words
* Sentence structure

Then, we can use this knowledge for tasks such as:

* Sentiment Analysis
* Text Classification
* Named Entity Recognition (NER)
* Question Answering
* Text Generation
* Translation
* Text Summarization

---

## What Is a Transformer?

A **Transformer** is a neural network architecture designed to understand relationships between words in a sequence.

One of its most important ideas is **Attention**.

### Example

Consider:

> "The bank is near the river."

The model needs to understand that **bank** refers to a place near a river.

Now:

> "I deposited money in the bank."

Here, **bank** has a different meaning.

Transformers use **context and attention** to understand these differences.

---

## Popular Pre-trained Transformer Models

| Model          | Main Use                                   |
| -------------- | ------------------------------------------ |
| **BERT**       | Text understanding and classification      |
| **RoBERTa**    | Improved BERT-style language understanding |
| **DistilBERT** | Smaller and faster version of BERT         |
| **GPT**        | Text generation                            |
| **T5**         | Text-to-text tasks                         |
| **BART**       | Text generation and summarization          |

---

## Example: BERT for Sentiment Analysis

Instead of creating and training our own NLP model, we can use a pre-trained BERT model that has already learned from a large dataset.

Conceptually:

```text
Text
  ↓
Tokenizer
  ↓
Pre-trained Transformer
  ↓
Sentiment Classification
  ↓
Positive / Negative
```

For example:

```text
"I really enjoyed this movie!"
              ↓
        BERT Model
              ↓
          POSITIVE
```

---

## Pre-trained Model vs Traditional Approach

### Traditional Machine Learning

```text
Raw Text
   ↓
Preprocessing
   ↓
TF-IDF / Bag of Words
   ↓
Machine Learning Model
   ↓
Prediction
```

### Pre-trained Transformer

```text
Raw Text
   ↓
Tokenizer
   ↓
Pre-trained Transformer
   ↓
Prediction
```

The transformer has already learned a large amount of language information during **pre-training**.

---

## Fine-Tuning

A pre-trained model can also be **fine-tuned** for a specific task.

For example:

```text
Pre-trained BERT
       ↓
Fine-tune on movie reviews
       ↓
Movie Sentiment Model
```

The model starts with general language knowledge and then learns the specific task from your dataset.

---

## Why Use Pre-trained Transformers?

### Advantages

* **Less training time**
* **Less data required**
* **Better performance** on many NLP tasks
* Can be adapted to different NLP problems
* Already contains useful language representations

### Simple Definition

> **A pre-trained Transformer is an NLP model that has already learned language patterns from large amounts of text and can be reused or fine-tuned for specific NLP tasks.**


# Sentiment Analysis Pipeline

```python
sentiment_pipeline = pipeline("sentiment-analysis")
```

This line creates a **pre-trained sentiment analysis pipeline** using the **Hugging Face Transformers** library.

## What is `pipeline()`?

`pipeline()` is a high-level function that allows us to use a **pre-trained Transformer model** without manually handling all the steps.

It automatically takes care of:

```text
Text
  ↓
Tokenizer
  ↓
Pre-trained Transformer Model
  ↓
Sentiment Prediction
```

## What does `"sentiment-analysis"` mean?

```python
pipeline("sentiment-analysis")
```

tells Transformers:

> "I want to perform Sentiment Analysis."

The pipeline automatically loads a suitable **pre-trained model** and its **tokenizer**.

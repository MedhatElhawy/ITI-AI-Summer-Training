### Stopwords

```python
# Stop words are common words that usually don't carry much meaning, such as: "and", "of", "a", "to".

# We remove them because:

# They reduce data complexity.
# They create a smaller and cleaner dataset.
# They can improve machine learning accuracy.
# They can make text processing faster.
```

```python
# import nltk library
import nltk

# download the stopwords dataset
nltk.download('stopwords')

# import stopwords from nltk corpus
# stopwords are common words that usually do not carry important meaning
from nltk.corpus import stopwords
```

```python
# assign our stop words to a variable
en_stopwords = stopwords.words('english')
```

```python
# print the list of stop words to see what we will be removing
print(en_stopwords)
```

```python
sentence = "it was too far to go to the shop and he did not want her to walk"
```

```python
# keep the words in the sentance if the word is not in the list of stop words
sentance_no_stopwords = ' '.join([word for word in sentence.split() if word not in (en_stopwords)])
print(sentance_no_stopwords)
```

```python
# removing stop words from list
en_stopwords.remove("did")
en_stopwords.remove("not")
```

```python
# add custom stop words
en_stopwords.append("go")
```

```python
sentance_no_stopwords_custom = ' '.join([word for word in sentence.split() if word not in (en_stopwords)])
print(sentance_no_stopwords_custom)
```

### Regular Expressions

```python
# regular expressions (regex) are used to search and find text that matches a specific pattern.
# regex is useful for filtering and cleaning text based on patterns instead of exact strings.
```

regular expressions (regex) are used to search, find, and manipulate text based on specific patterns.

## 1. Regex Syntax

| Regex | Meaning | Example |
|:---:|---|---|
| `.` | any character except newline | `c.t` → `cat` |
| `[]` | character set | `[abc]` → `a`, `b`, `c` |
| `[^...]` | anything except the characters inside | `[^0-9]` → letters |
| `[a-z]` | lowercase letters | `[a-z]+` |
| `[A-Z]` | uppercase letters | `[A-Z]+` |
| `[0-9]` | digits | `[0-9]+` |
| `\d` | any digit | `\d+` → `123` |
| `\D` | any non-digit | `\D+` → `abc` |
| `\w` | word character | `\w+` → `hello123` |
| `\W` | non-word character | `\W+` → `!@#` |
| `\s` | whitespace | `\s+` |
| `\S` | non-whitespace | `\S+` |

---

## 2. Quantifiers

| Regex | Meaning | Example |
|:---:|---|---|
| `*` | 0 or more | `a*` |
| `+` | 1 or more | `a+` |
| `?` | 0 or 1 | `a?` |
| `{n}` | exactly n times | `a{3}` |
| `{n,}` | n or more times | `a{2,}` |
| `{n,m}` | between n and m times | `a{2,4}` |

---

## 3. Anchors

| Regex | Meaning | Example |
|:---:|---|---|
| `^` | beginning of string | `^hello` |
| `$` | end of string | `world$` |
| `\b` | word boundary | `\bcat\b` |
| `\B` | not a word boundary | `\Bcat` |

---

## 4. Groups and Operators

| Regex | Meaning | Example |
|:---:|---|---|
| `()` | capturing group | `(cat\|dog)` |
| `(?:...)` | non-capturing group | `(?:cat\|dog)` |
| `\|` | OR | `cat\|dog` |
| `\` | escape special character | `\.` |
| `\.` | literal period | `\.` |
| `\?` | literal question mark | `\?` |
| `\+` | literal plus sign | `\+` |
| `\*` | literal asterisk | `\*` |

---

## 5. Common NLP Patterns

| Pattern | Purpose |
|:---:|---|
| `r"\d+"` | find numbers |
| `r"\w+"` | find words |
| `r"\s+"` | find whitespace |
| `r"[^\w\s]"` | find punctuation and special characters |
| `r"https?://\S+"` | find URLs |
| `r"[\w\.-]+@[\w\.-]+\.\w+"` | find email addresses |

---

## 6. Python `re` Functions

| Function | Purpose |
|---|---|
| `re.search()` | find the first match |
| `re.match()` | check the beginning of a string |
| `re.findall()` | return all matches as a list |
| `re.finditer()` | return all matches as match objects |
| `re.sub()` | replace matched text |
| `re.split()` | split text using a regex pattern |
| `re.fullmatch()` | check if the entire string matches |

---

#### re.search

```python
# `re.search()` is used to check if a specific pattern exists anywhere in a string.

# **syntax:**

# `re.search("pattern", "string")`

# - if the pattern is found, it returns a **match object**.
# - if the pattern is not found, it returns **None**.
```

```python
result_search = re.search("pattern", r"string containing the pattern")
print(result_search)
```

```python
print(result_search.group()) # returns just the matching pattern
```

```python
result_search = re.search("pattern",r"the phrase to find isn't in this string")
print(result_search) # returns None
```

#### re.sub

```python
# `re.sub()` is used to find a specific pattern in a string and replace it with another value.

# **syntax:**

# `re.sub("pattern", "replacement", "string")`

# - finds the specified pattern.
# - replaces the matched text with the replacement text.
# - returns the updated string.
```

```python
string = r"sara was able to help me find the items i needed quickly"
```

```python
new_string = re.sub(r"sara", r"sarah", string) # replace the incorrect spelling of sarah
print(new_string)
```

### Regex Syntax

```python
customer_reviews = ['sam was a great help to me in the store', 
                    'the cashier was very rude to me, I think her name was eleanor', 
                    'amazing work from sadeen!', 
                    'sarah was able to help me find the items i needed quickly', 
                    'lucy is such a great addition to the team', 
                    'great service from sara she found me what i wanted'
                   ]
```

```python
#Find only sarah's reviews but account for the spelling of sara
sarahs_reviews = []
pattern_to_find = r"sarah?" 
# the ? after r means it is an optional character to match, so our search will look for sarah and sara
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        sarahs_reviews.append(string)
```

```python
print(sarahs_reviews)
```

```python
#Find reviews that start with the letter a
a_reviews = []
pattern_to_find = r"^a" # the ^ operator to indicates the start of a string
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        a_reviews.append(string)
```

```python
print(a_reviews)
```

```python
#Find reviews that end with the letter y
y_reviews = []
pattern_to_find = r"y$" # the $ operator to indicate the end of a string
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        y_reviews.append(string)
```

```python
print(y_reviews)
```

```python
#Find reviews that contain the words needed or wanted
needwant_reviews = []
pattern_to_find = r"(need|want)ed" # the pipe operator | can be used to mean OR
```

```python
for string in customer_reviews:
    if (re.search(pattern_to_find, string)):
        needwant_reviews.append(string)
```

```python
print(needwant_reviews)
```

```python
#Remove anything from the review that isn't a word or a space (i.e. remove punctuation)
no_punct_reviews = []
pattern_to_find = r"[^\w\s]" 
# [^ ] means "not", \w means word and \s means whitespace: so find anything that is not a word or a space
```

```python
for string in customer_reviews:
    no_punct_string = re.sub(pattern_to_find, "", string)
    no_punct_reviews.append(no_punct_string)
```

```python
print(no_punct_reviews)
```

### Tokenizing Text

tokenization is the process of splitting text into smaller units called **tokens**.

common types of tokenization include:
- **word tokenization** → splits text into individual words.
- **sentence tokenization** → splits text into sentences.
- **subword tokenization** → splits words into smaller parts.
- **character tokenization** → splits text into individual characters.

### why do we use tokenization?

- it helps us analyze and understand the individual parts of a text.
- it makes the overall meaning of the text easier to analyze.
- it is an important step before **vectorization**.
- nltk can be used for sentence and word tokenization.

```python
import nltk
nltk.download('punkt_tab')
from nltk.tokenize import word_tokenize, sent_tokenize
```

```python
#Sentance tokenization
sentences = "Her cat's name is Luna. Her dog's name is max"
sent_tokenize(sentences)
```

```python
#Word tokenization
sentence = "Her cat's name is Luna"
word_tokenize(sentence)
```

```python
sentence_2 = "Her cat's name is Luna and her dog's name is max"
word_tokenize(sentence_2)
```

### Stemming

stemming is a text preprocessing technique used to reduce words to their base or root form.

for example:
- connecting → connect
- connected → connect
- playing → play

### why do we use stemming?

- it reduces the number of unique words.
- it makes the dataset smaller and less complex.
- it helps reduce noise in the text.
- it prepares the data for machine learning.

### note

stemming removes word endings or suffixes, so the result may sometimes not be a meaningful or valid word.

```python
from nltk.stem import PorterStemmer
```

```python
# create stemmer
ps = PorterStemmer()
```

```python
connect_tokens = ['connecting', 'connected', 'connectivity', 'connect', 'connects']
```

```python
for t in connect_tokens:
    print(t, " : ", ps.stem(t))
```

```python
learn_tokens = ['learned', 'learning', 'learn', 'learns', 'learner', 'learners']
```

```python
for t in learn_tokens:
    print(t, " : ", ps.stem(t))
```

```python
likes_tokens = ['likes', 'better', 'worse']
```

```python
for t in likes_tokens:
    print(t, " : ", ps.stem(t))
```

### Lemmatization

lemmatization is a text preprocessing technique that reduces words to their meaningful base form, called a **lemma**.

unlike stemming, lemmatization uses a dictionary and the context of the word to find the correct base form.

### stemming vs lemmatization

- **stemming** → removes word endings, which may produce an invalid word.
- **lemmatization** → finds the meaningful base form while preserving the word's meaning.

### example

- better → good
- running → run
- studies → study

lemmatization is more accurate than stemming but can be slower because it uses linguistic information and context.

```python
# import the nltk library
import nltk

# download the wordnet dictionary used for lemmatization
nltk.download('wordnet')

# import the wordnet lemmatizer
from nltk.stem import WordNetLemmatizer
```

```python
# create lemmatizer 
lemmatizer = WordNetLemmatizer()
```

```python
for t in connect_tokens:
    print(t, " : ", lemmatizer.lemmatize(t))
```

```python
for t in learn_tokens:
    print(t, " : ", lemmatizer.lemmatize(t))
```

```python
for t in likes_tokens:
    print(t, " : ", lemmatizer.lemmatize(t))
```

### N-grams

## n-grams

an **n-gram** is a sequence of `n` neighboring words (or tokens) in a text.

n can be any number:

- **unigram (n=1)** → one word
- **bigram (n=2)** → two consecutive words
- **trigram (n=3)** → three consecutive words

### why do we use n-grams?

- to check and evaluate our text preprocessing.
- to explore the content of the data.
- to capture relationships between neighboring words.
- to create new features for machine learning.

### example

text: `"i love natural language processing"`

- unigram → `i`, `love`, `natural`, `language`, `processing`
- bigram → `i love`, `love natural`, `natural language`, `language processing`
- trigram → `i love natural`, `love natural language`, `natural language processing`

```python
import nltk
import pandas as pd
import matplotlib.pyplot as plt
```

```python
tokens = ['the', 'rise', 'of', 'artificial', 'intelligence', 'has', 'led', 'to', 'significant', 'advancements', 'in', 'natural', 'language', 'processing', 'computer', 'vision', 'and', 'other', 'fields', 'machine', 'learning', 'algorithms', 'are', 'becoming', 'more', 'sophisticated', 'enabling', 'computers', 'to', 'perform', 'complex', 'tasks', 'that', 'were', 'once', 'thought', 'to', 'be', 'the', 'exclusive', 'domain', 'of', 'humans', 'with', 'the', 'advent', 'of', 'deep', 'learning', 'neural', 'networks', 'have', 'become', 'even', 'more', 'powerful', 'capable', 'of', 'processing', 'vast', 'amounts', 'of', 'data', 'and', 'learning', 'from', 'it', 'in', 'ways', 'that', 'were', 'not', 'possible', 'before', 'as', 'a', 'result', 'ai', 'is', 'increasingly', 'being', 'used', 'in', 'a', 'wide', 'range', 'of', 'industries', 'from', 'healthcare', 'to', 'finance', 'to', 'transportation', 'and', 'its', 'impact', 'is', 'only', 'set', 'to', 'grow', 'in', 'the', 'years', 'to', 'come']
print(tokens)
```

```python
# unigrams: n=1
unigrams = (pd.Series(nltk.ngrams(tokens, 1)).value_counts()) 
print(unigrams)
```

```python
unigrams[0:10].sort_values().plot.barh(color='lightsalmon', width=.9, figsize=(12, 9))
plt.title('10 Most Frequently Occuring Unigrams')
plt.ylabel('Unigram')
plt.xlabel('# of Occurances')
```

```python
# bigrams: n=2
bigrams = (pd.Series(nltk.ngrams(tokens, 2)).value_counts()) 
print(bigrams)
```

```python
bigrams[0:10].sort_values().plot.barh(color='skyblue', width=.9, figsize=(12, 8))
plt.title('10 Most Frequently Occuring Bigrams')
plt.ylabel('Bigram')
plt.xlabel('# of Occurances')
```

```python
# trigrams: n=3
trigrams = (pd.Series(nltk.ngrams(tokens, 3)).value_counts()) 
print(trigrams)
```
